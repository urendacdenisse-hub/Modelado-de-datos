## Bases de datos

Conectémonos a una base de datos SQLite y ejecutemos una consulta para cargar los resultados en un DataFrame.

**Librería:** `sqlite3`

`sqlite3` permite conectarse y ejecutar consultas sobre bases de datos SQLite desde Python.

**Funciones utilizadas**

- `sqlite3.connect()`: Abre una conexión con una base de datos SQLite.
- `cursor()`: Crea un cursor para ejecutar instrucciones SQL.
- `execute()`: Ejecuta una instrucción SQL.
- `fetchall()`: Recupera todos los resultados de la consulta.

```python
import sqlite3

conn = sqlite3.connect("modelado-de-datos/datos/chinook.sqlite")
cursor = conn.cursor()

cursor.execute(
    "SELECT name FROM sqlite_master WHERE type='table';"
)
print(cursor.fetchall())
```

**Funciones utilizadas**

- `pd.read_sql_query()`: Ejecuta una consulta SQL y devuelve el resultado como un DataFrame.


```python
artists = pd.read_sql_query(
    "SELECT * FROM Artist;",
    conn
)

artists.head()
```

**Funciones utilizadas**

- `close()`: Cierra la conexión con la base de datos y libera los recursos asociados.

```python
conn.close()
```

## APIs

Enviemos una solicitud a una API y recuperemos la información en formato JSON.

**Librería:** `requests`

`requests` permite enviar solicitudes a servicios web y APIs desde Python.

**Funciones utilizadas**

- `requests.get()`: Solicita información desde la URL especificada.
- `json()`: Convierte la respuesta de la API en un objeto de Python (diccionario o lista).

```python
import requests

url = (
    "https://api.open-meteo.com/v1/forecast"
    "?latitude=31.74&longitude=-106.49&current=temperature_2m"
)

response = requests.get(url)
data = response.json()
```
