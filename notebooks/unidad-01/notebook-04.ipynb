# Diagnóstico de los supuestos de regresión lineal

En esta lección analizaremos distintos conjuntos de datos ficticios diseñados para explorar los supuestos de la regresión lineal.

Comenzaremos con un escenario donde los supuestos se cumplen razonablemente bien y, posteriormente, analizaremos casos donde algunos de ellos son violados.


## Regresión lineal simple

### Dataset

`01_relaciones_mixtas`

### Variables

- **Predictora:** `horas_estudio`
- **Respuesta:** `puntaje_examen`

### ¿Qué debemos esperar?

El puntaje del examen tiene una relación lineal positiva con las horas de estudio.

Esperamos observar:

- una relación aproximadamente lineal;
- residuos aproximadamente normales;
- varianza aproximadamente constante de los residuos;
- observaciones independientes.


```python
import pandas as pd

ruta = "../../datos/datasets_regresion_lineal.xlsx"

df = pd.read_excel(ruta, sheet_name="01_relaciones_mixtas" )
df = df[["horas_estudio", "puntaje_examen"]]
df.head()
```

# Cuando los supuestos no se cumplen

Hasta ahora hemos trabajado con datos que se comportan de acuerdo con los supuestos del modelo lineal.

Ahora analizaremos qué ocurre cuando estos supuestos comienzan a fallar.

## No normalidad de los residuos

### Dataset

`03_no_normalidad`

### Variables

- **Predictoras:** `superficie_m2`, `antiguedad_anios`, `distancia_centro_km`, `indice_calidad_1_10`
- **Respuesta:** `precio_vivienda_miles`

### ¿Qué debemos esperar?

- La estructura del modelo continúa siendo lineal, pero los errores no siguen una distribución normal.

- Esperamos observar desviaciones de la normalidad al analizar los residuos del modelo.

```python
df = pd.read_excel(ruta, sheet_name="03_no_normalidad")
df.head()
```

## Heteroscedasticidad

### Dataset

`04_heteroscedasticidad`

### Variables

- **Predictoras:** `ingreso_mensual_miles`, `tamano_hogar`, `edad_jefe_hogar`
- **Respuesta:** `gasto_mensual_miles`

### ¿Qué debemos esperar?

- La relación entre los predictores y la respuesta continúa siendo aproximadamente lineal, pero la variabilidad de los errores no es constante.

- En particular, esperamos que la dispersión aumente conforme aumenta el nivel de ingreso.



```python
df = pd.read_excel(ruta, sheet_name="04_heteroscedasticidad")
df.head()
```

## Dependencia en los residuos

### Dataset

`05_dependencia_residuos`

### Variables

- **Predictoras:** `temperatura_c`, `ocupacion_pct`
- **Respuesta:** `consumo_energia_kwh`
- **Variable temporal:** `dia`

### ¿Qué debemos esperar?

En este conjunto de datos, las observaciones corresponden a mediciones realizadas a lo largo del tiempo.

Aunque la relación entre los predictores y la respuesta es lineal, los errores de observaciones consecutivas están relacionados entre sí.

Esperamos observar:

- una relación aproximadamente lineal;
- residuos aproximadamente normales;
- varianza aproximadamente constante;
- dependencia entre residuos consecutivos;
- un patrón temporal en los residuos.

Este comportamiento se conoce como autocorrelación.

```python
df = pd.read_excel(ruta, sheet_name="05_dependencia_residuos")
df.head()
```

### Dataset

### ¿Qué debemos esperar?

Algunos predictores contienen información muy similar.

En particular, variables relacionadas con el tamaño de la vivienda presentan una fuerte asociación entre sí.

Esperamos encontrar:

- correlaciones altas entre algunos predictores;
- dificultad para separar sus efectos individuales;
- posibles problemas en la estimación e interpretación de los coeficientes.

```python
df = pd.read_excel(ruta, sheet_name="06_multicolinealidad")
df.head()
```
