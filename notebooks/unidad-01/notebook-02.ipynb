## Exploración inicial

Carguemos un conjunto de datos y realicemos una exploración inicial de su contenido.

```python
import pandas as pd

df = pd.read_csv("modelado-de-datos/datos/sales.csv")
```

**Librería:** `pandas`

`pandas` proporciona herramientas para explorar y resumir la información contenida en un DataFrame.

**Funciones utilizadas**

- `info()`: Muestra la estructura del DataFrame, incluyendo el tipo de datos y los valores no nulos de cada columna.
- `shape`: Devuelve el número de filas y columnas del DataFrame.
- `isna()`: Identifica los valores faltantes en el DataFrame.
- `sum()`: Suma los valores de cada columna.
- `describe()`: Calcula estadísticas descriptivas de las variables numéricas.

```python
df.info()
```

```python
df.shape
```

```python
df.isna().sum()
```

```python
df.describe()
```

## Limpieza de datos

Eliminemos algunos problemas comunes antes de comenzar el análisis del conjunto de datos.

**Funciones utilizadas**

- `drop_duplicates()`: Elimina los registros duplicados.
- `fillna()`: Reemplaza los valores faltantes por un valor especificado.
- `dropna()`: Elimina los registros que contienen valores faltantes.
- `clip()`: Limita los valores de una columna a un intervalo definido.

```python
df.drop_duplicates()
```

```python
df.fillna(0)
```

```python
df.dropna()
```

```python
df["unit_price"].clip(lower=0)
```

## Transformación de datos

Transformemos las variables del conjunto de datos para prepararlas para su análisis.

**Funciones utilizadas**

- `astype()`: Convierte una columna al tipo de dato especificado.
- `mean()`: Calcula la media de una columna.
- `std()`: Calcula la desviación estándar de una columna.
- `pd.to_datetime()`: Convierte una columna al tipo de dato fecha y hora (`datetime`).

```python
df.info()
```

```python
df["quantity"].astype(int)
```

```python
(df["unit_price"] - df["unit_price"].mean()) / df["unit_price"].std()
```

```python
df["quantity"] * df["unit_price"]
```

```python
pd.to_datetime(df["date"])
```
